```python
A, B, C = map(int, input().split())
if (A > B) and (A > C):
    print("A is greater!")
elif (B > A) and (B > C):
    print("B is greater!")
else:
    print("C is greater!")
```

## Nested if is used when order of execution is priority

```python
a, b = True, True

if a:
    if b:
        print("Both are satisfied!")
```

```python
N = int(input())
if N >= 1 and N <= 10:
    print("Try Next Time")
elif N >= 11 and N <= 20:
    print("Variable")
elif N >= 21 and N <= 25:
    print("Good Person")
elif N >= 26 and N <= 30 and N != 28:
    print("Better luck next time")
```
