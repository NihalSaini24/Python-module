```python
def is_perfect_number(n):
    if n <= 1:
        return false
    total, i = 1,2
    while i * i <= n:
        if n % i == 0:
            total += i
            if i != n // i:
                total += n // i
        i += 1
    return total == n
print(is_perfect_number(496))
```

```python
def prime_factors(n): 
    res = [] 
    while n % 2 == 0: 
        res.append(2)
        n //= 2 
    i = 3 
    while i * i <= n: 
        while n % i == 0: 
            res.append(i)
            n //= i 
        i += 2 
    if n > 1: 
        res.append(n) 
    return res
print(prime_factors(63))
```

```python
def automorphic_num(n):
    return str(n * n).endswith(str(n))

print(automorphic_num(19))
```

```python
def trailing_zeroes(n):
    c = 0
    while n:
        n //= 5
        c += n
    return c

print(trailing_zeroes(5))
```

```python
def sum_of_dig(n):
    tot = 0
    while(n>0):
        digit = n % 10
        tot += digit
        n = n // 10
    return tot
# print(sum_of_dig(41))
def harshad_num(n):
    # s = sum(int(d) for d in str(n))
    # print(s)
    s = sum_of_dig(n)
    return n % s == 0
print(harshad_num(41))
```

```python
def harmonic_sum(n):
    # return sum(1 / i for i in range(1, n + 1))
    total = 0
    for i in range(1, n+1):
        total += 1 / i
    return total
print(harmonic_sum(10))
```

```python
def to_binary(n):
    if n == 0:
        return "0"
    bits = []
    while n:
        bits.append(str(n % 2))
        n //= 2
    return "".join(bits[::-1])

print(to_binary(10))
```

```python
def pairs_sum(a, target):
    # return [(a[i],a[j]) for i in range(len(a))
    #                                  for j in range(i+1,len(a))
    #                                  if(a[i] + a[j] == target)]
    for i in range(len(a)):
        for j in range(i+1,len(a)):
            if(a[i] + a[j] == target):
                return(a[i],a[j])
    
print(pairs_sum([1,2,3,4,5,6,7,8,9], 5))
```
