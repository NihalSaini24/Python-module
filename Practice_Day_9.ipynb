## For Loop

```python
for i in range(3):
    for j in range(4):
        for k in range(5):
            print(i,j,k)
```

```python
ind = [(19,41),(11,19),(41,56),(56,41)]
for i in ind:
    print(i)
```

```python
ind = [(19,41),(11,19),(41,56),(56,41)]
for i in ind:
    print(i[0],end = " ")
    print(i[1])
```

```python
ind = [(19,41),(11,19),(41,56),(56,41)]
for i,j in ind:
    print(i,j)
```

## Lists

```python
arr = [x for x in range(10)]
print(arr[0:5])
```

```python
str = "Nihal"
print(str[::-1])
```

```python
arr = [x for x in range(10)]
print(arr)
dup = []
dup.append(arr[:])
print(dup)
```

```python
print([x for x in range(10) if x % 2 == 0])
```
