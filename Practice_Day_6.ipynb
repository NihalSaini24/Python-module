Conditional Statements

```python
N = int(input())
if (N % 2 == 0):
    print("Even")
else:
    print("Odd")
```

```python
N = int(input())
if N != 0:
    if (N % 2 == 0):
        print("Even")
    else:
        print("Odd")
```

```python
N = int(input())
if N != 0:
    if (N % 2 == 0) and (N % 3 == 0):
        print("Both")
    elif (N % 2 == 0) or (N % 3 == 0):
        print("Anyone")
    else:
        print("Not")
```
