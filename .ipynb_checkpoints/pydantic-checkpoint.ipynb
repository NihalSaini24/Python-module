```python
from pydantic import BaseModel

class Patient(BaseModel):
    name: str
    age: int

# Valid data
p1 = Patient(name="John", age=30)
print("Success:", p1)

# Invalid data (will raise an error because age is a string)
# p2 = Patient(name="Jane", age="thirty")
```

```python
from pydantic import BaseModel

class Patient(BaseModel):
    name: str
    city: str = "Mumbai" # Default value

p1 = Patient(name="John")
print(p1) # City automatically becomes Mumbai
```

```python
from pydantic import BaseModel
from typing import Optional

class Patient(BaseModel):
    name: str
    middle_name: Optional[str] = None

# Valid: No middle name provided
p1 = Patient(name="John")
print(p1)
```

```python
from pydantic import BaseModel
from typing import List

class Patient(BaseModel):
    name: str
    allergies: List[str]

p1 = Patient(name="John", allergies=["Peanuts", "Dust"])
print(p1)

# Invalid: If you pass [123, "Dust"], Pydantic will throw an error.
```

```python
from pydantic import BaseModel
from typing import Dict

class Patient(BaseModel):
    name: str
    test_scores: Dict[str, int]

p1 = Patient(name="John", test_scores={"Blood Pressure": 120, "Sugar": 90})
print(p1)
```

```python
from pydantic import BaseModel, Field

class Patient(BaseModel):
    name: str
    # Age must be between 0 and 120
    age: int = Field(gt=0, lt=120, description="Age in years")

p1 = Patient(name="John", age=30)
print(p1)

# p2 = Patient(name="Jane", age=150) # This will FAIL
```

```python
from pydantic import BaseModel, Field
from typing import Annotated

class Patient(BaseModel):
    # Sticky note on the 'str' type
    name: Annotated[str, Field(min_length=2, description="Full name")]
    age: int = Field(gt=0, lt=120)

p1 = Patient(name="John", age=30)
print(p1)
```

```python
from pydantic import BaseModel, field_validator

class Patient(BaseModel):
    name: str

    @field_validator('name')
    @classmethod
    def uppercase_name(cls, value):
        # Change "john" to "JOHN"
        return value.upper()

p1 = Patient(name="john")
print(p1)
```

```python
from pydantic import BaseModel, field_validator

class Patient(BaseModel):
    name: str
    email: str

    @field_validator('email')
    @classmethod
    def check_hospital_email(cls, value):
        if not value.endswith("@hospital.com"):
            raise ValueError("Email must be a hospital email!")
        return value

p1 = Patient(name="John", email="john@hospital.com")
print("Success:", p1)

# try:
#     p2 = Patient(name="Jane", email="jane@gmail.com") # This will FAIL
# except ValueError as e:
#     print("Failed:", e)
```

```python
from pydantic import BaseModel, Field, field_validator
from typing import Optional, List, Dict, Annotated

class HospitalPatient(BaseModel):
    # 1. Annotated with Field constraints
    name: Annotated[str, Field(min_length=2, max_length=50)]

    # 2. Optional field
    middle_name: Optional[str] = None

    # 3. Field with numeric constraints
    age: Annotated[int, Field(gt=0, lt=120)]

    # 4. List of strings
    allergies: List[str] = []

    # 5. Dictionary
    test_scores: Dict[str, int] = {}

    # 6. Field for custom validation
    email: str

    # 7. Field Validator for custom logic
    @field_validator('email')
    @classmethod
    def validate_email(cls, value):
        if not value.endswith("@hospital.com"):
            raise ValueError("Email must end with @hospital.com")
        return value

# --- TESTING ---
# Valid Patient
p1 = HospitalPatient(
    name="John Doe",
    age=30,
    email="john@hospital.com",
    allergies=["Peanuts"],
    test_scores={"Blood": 95}
)
print("Valid Data:\n", p1)

# Invalid Patient (Uncomment to see the error)
# p2 = HospitalPatient(
#     name="J", # Too short
#     age=150, # Too old
#     email="jane@gmail.com", # Wrong domain
#     allergies=["Dust"],
#     test_scores={"Sugar": 90}
# )
```
