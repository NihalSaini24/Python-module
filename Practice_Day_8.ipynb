## Loops

```python
T = int(input())
while (T > 0):
    T = T - 1
    N = int(input())
    arr = list(map(int, input().split()))
    total = 0
    for i in range(N):
        total += arr[i]
    print(total)
```

```python
N = int(input())
total = 0
while (N > 0):
    total += N
    N = N - 1
print(total)
```
